```python
def create_simple_grid(size=SIZE):
    grid = [[0] * size for _ in range(size)]
    num_rect = 2

    rectangles = []
    for _ in range(num_rect):
        rect_width = random.randint(20, size // 3) 
        rect_height = random.randint(20, size // 3) 
        
        rect_x = random.randint(10, size - rect_width - 10)  
        rect_y = random.randint(10, size - rect_height - 10) 
            
        rectangles.append((rect_x, rect_y, rect_width, rect_height))
        
        for y in range(rect_y, rect_y + rect_height):
            for x in range(rect_x, rect_x + rect_width):
                grid[y][x] = 1  
                
    agent_x = random.randint(size // 2, size - size // 4)
    agent_y = random.randint(1, size // 4)
    

    while grid[agent_x][agent_y] == 1: 
        agent_x = random.randint(size // 2, size - size // 4)
        agent_y = random.randint(1, size // 4)
  
    goal_x = random.randint(size // 2, size - size // 3)
    goal_y = random.randint(size - size // 6, size - 2)
    

    while grid[goal_x][goal_y] == 1:  
        goal_x = random.randint(size // 2, size - size // 3)
        goal_y = random.randint(size - size // 6, size - 2)

    
    return {
        "name": "simple",
        "agent_x": agent_x,
        "agent_y": agent_y,
        "goal_x": goal_x,
        "goal_y": goal_y,
        "size": size,
        "grid": grid
    }
```

```python
def create_star_grid(size=SIZE):
    grid = [[0] * size for _ in range(size)]
    
    # unten rechts
    l1_x, l1_y = size - size // 4, size - size // 3
    l1_length = size // 4             
    l1_thickness = size // 10                  
    
    # Horizontal
    for y in range(l1_y, l1_y + l1_thickness):
        for x in range(l1_x, l1_x + l1_length):
            grid[y][x] = 1
    
    # Vertikal
    for y in range(l1_y, l1_y + l1_length):
        for x in range(l1_x, l1_x + l1_thickness):
            grid[y][x] = 1

    # oben links
    l2_x, l2_y = 0, size // 4
    l2_length = size // 4 
    l2_thickness = size // 10
    
    # Horizontal
    for y in range(l2_y, l2_y + l2_thickness):
        for x in range(l2_x, l2_x + l2_length):
            grid[y][x] = 1
    
    # Vertikal
    for y in range(l2_y, l2_y - l2_length + 10, -1):  # y geht nach oben -1, 10 ist ne lücke
        for x in range(l2_x + l2_length - l2_thickness, l2_x + l2_length):
            grid[y][x] = 1

    
    agent_x = random.randint(1, size // 4)
    agent_y = random.randint(1, size // 4)
   
    while grid[agent_x][agent_y] == 1: 
        agent_x = random.randint(1, size // 4)
        agent_y = random.randint(1, size // 4)
    
 
    goal_x = random.randint(size - size // 6, size - 2)
    goal_y = random.randint(size - size // 6, size - 2)
    
  
    while grid[goal_x][goal_y] == 1:  
        goal_x = random.randint(size - size // 6, size - 2)
        goal_y = random.randint(size - size // 6, size - 2)

    return {
        "name": "star",
        "agent_x": agent_x,
        "agent_y": agent_y,
        "goal_x": goal_x,
        "goal_y": goal_y,
        "size": size,
        "grid": grid
    }
```

```python
def create_bottleneck_grid(size=SIZE):
    grid = [[0] * size for _ in range(size)]
    
    # blocked balken 
    balken_y = size // 3
    balken_dicke = 50
    
    for x in range(size):
        for y in range(balken_y, balken_y + balken_dicke):
            grid[y][x] = 1 
    
    # unblock path
    weg_breite = 5
    weg_x = size // 2  
    
    for x in range(weg_x - weg_breite // 2, weg_x + weg_breite // 2):
        for y in range(size):
            grid[y][x] = 0 
    
    
    agent_x = random.randint(size // 2, size - size // 4)
    agent_y = random.randint(1, size // 4)
    

    while grid[agent_x][agent_y] == 1:  
        agent_x = random.randint(size // 2, size - size // 4)
        agent_y = random.randint(1, size // 4)
    
    goal_x = random.randint(size // 2, size - size // 4)
    goal_y = random.randint(size - size // 6, size - 2)
    
    
    while grid[goal_x][goal_y] == 1:  
        goal_x = random.randint(size // 2, size - size // 4)
        goal_y = random.randint(size - size // 6, size - 2)

    
    return {
        "name": "bottleneck",
        "agent_x": agent_x,
        "agent_y": agent_y,
        "goal_x": goal_x,
        "goal_y": goal_y,
        "size": size,
        "grid": grid
    }
```

```python
def create_trap_grid(size=SIZE):
    grid = [[0] * size for _ in range(size)]

    rect_start_x = size // 5
    rect_start_y = size // 5
    rect_end_x = size - size // 5
    rect_end_y = size - size // 5
        
    for i in range(rect_start_x, rect_end_x):
        for j in range(rect_start_y, rect_end_y):
            grid[i][j] = 1 

    hole_start_x = rect_start_x
    hole_end_x = hole_start_x + random.randint(rect_end_x // 4, rect_end_x // 2)
    hole_start_y = rect_start_y + random.randint(10, (rect_end_y - rect_start_y) // 4)
    hole_end_y = hole_start_y + random.randint(rect_end_y // 4, rect_end_y // 2)

    for i in range(hole_start_x, hole_end_x):
        for j in range(hole_start_y, hole_end_y):
            if random.random() > 0.1: 
                grid[i][j] = 0  
                
   
    agent_x = random.randint(size // 2, size - size // 4)
    agent_y = random.randint(1, size // 4)
    
   
    while grid[agent_x][agent_y] == 1:  
        agent_x = random.randint(size // 2, size - size // 4)
        agent_y = random.randint(1, size // 4)
    
 
    goal_x = random.randint(size // 2, size - size // 4)
    goal_y = random.randint(size - size // 6, size - 2)
    
    while grid[goal_x][goal_y] == 1:  
        goal_x = random.randint(size // 2, size - size // 4)
        goal_y = random.randint(size - size // 6, size - 2)

    
    return {
        "name": "trap",
        "agent_x": agent_x,
        "agent_y": agent_y,
        "goal_x": goal_x,
        "goal_y": goal_y,
        "size": size,
        "grid": grid
    }
```

```python
def create_doubletrap_grid(size=SIZE):
    grid = [[0] * size for _ in range(size)]

    rect_start_x = size // 5
    rect_start_y= size // 5
    rect_end_x = size - size // 5
    rect_end_y = size - size // 5

    # trap 1
    for i in range(rect_start_x, rect_end_x):
        for j in range(rect_start_y, rect_end_y):
            grid[i][j] = 1  

    hole_start_x = rect_start_x
    hole_end_x = hole_start_x + random.randint(rect_end_x // 5, rect_end_x // 3)
    hole_start_y = rect_start_y + random.randint(10, (rect_end_y - rect_start_y) // 4)
    hole_end_y = hole_start_y + random.randint(rect_end_y // 4, rect_end_y // 2)

    for i in range(hole_start_x, hole_end_x):
        for j in range(hole_start_y, hole_end_y):
            if random.random() > 0.1: 
                grid[i][j] = 0  
                
    # trap 2        
    hole_start_x_mirror = size - hole_end_x
    hole_end_x_mirror = size - hole_start_x
    

    hole_start_y_mirror = size - hole_end_y
    hole_end_y_mirror = size - hole_start_y
    

    for i in range(hole_start_x_mirror, hole_end_x_mirror):
        for j in range(hole_start_y_mirror, hole_end_y_mirror):
            if random.random() > 0.1:  
                grid[i][j] = 0


    agent_x = random.randint(size // 2, size - size // 4)
    agent_y = random.randint(1, size // 4)
    
    while grid[agent_x][agent_y] == 1:  
        agent_x = random.randint(size // 2, size - size // 4)
        agent_y = random.randint(1, size // 4)
    

    goal_x = random.randint(size // 2, size - size // 4)
    goal_y = random.randint(size - size // 6, size - 2)
    

    while grid[goal_x][goal_y] == 1:  
        goal_x = random.randint(size // 2, size - size // 4)
        goal_y = random.randint(size - size // 6, size - 2)

    
    return {
        "name": "doubletrap",
        "agent_x": agent_x,
        "agent_y": agent_y,
        "goal_x": goal_x,
        "goal_y": goal_y,
        "size": size,
        "grid": grid
    }
```

```python
def create_detour_grid(size=SIZE):
    grid = [[0] * size for _ in range(size)]

    rect_start_x = size // 5
    rect_start_y = 0
    rect_end_x = size - size // 5 # länge nach unten
    rect_end_y = size - size // 5

    # trap 1
    for i in range(rect_start_x, rect_end_x):
        for j in range(rect_start_y, rect_end_y):
            grid[i][j] = 1  

    hole_start_x = rect_start_x
    hole_end_x = rect_end_y
    hole_start_y = rect_start_y + random.randint(10, (rect_end_y - rect_start_y) // 4)
    hole_end_y = hole_start_y + random.randint(rect_end_y // 4, rect_end_y // 2)

    for i in range(hole_start_x, hole_end_x):
        for j in range(hole_start_y, hole_end_y):
            if random.random() > 0.4:  
                grid[i][j] = 0  

 
    agent_x = random.randint(1, size // 4)
    agent_y = random.randint(1, size // 4)
    
    while grid[agent_x][agent_y] == 1:  
        agent_x = random.randint(1, size // 4)
        agent_y = random.randint(1, size // 4)
    
 
    goal_x = random.randint(1, size // 2)
    goal_y = random.randint(size - size // 4, size)
    
  
    while grid[goal_x][goal_y] == 1:  
        goal_x = random.randint(1, size // 2)
        goal_y = random.randint(size - size // 4, size)

    
    return {
        "name": "detour",
        "agent_x": agent_x,
        "agent_y": agent_y,
        "goal_x": goal_x,
        "goal_y": goal_y,
        "size": size,
        "grid": grid

    }
```

```python
def save_grid_to_json(grid, filename):
    with open(filename, 'w') as f:
        json.dump(grid, f, indent=2)


grids = [create_bottleneck_grid(), create_simple_grid(), create_star_grid(), 
         create_trap_grid(), create_doubletrap_grid(), create_detour_grid()]


for grid in grids:
    save_grid_to_json(grid, f"{grid['name']}_grid_128.json")
```
