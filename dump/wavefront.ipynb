```python
import json
```

```python
def wavefront_algorithm(grid, start, goal):
    rows, cols = len(grid), len(grid[0])
    distance = [[float('inf')] * cols for _ in range(rows)]
    
    sx, sy = start
    gx, gy = goal
    
    if grid[sx][sy] or grid[gx][gy]:
        raise ValueError("Start or goal pos blocked.")

    distance[sx][sy] = 0

    last_distance = 0
    new_entry = True

    while distance[gx][gy] == float('inf') and new_entry:
        new_entry = False

        for x in range(rows):
            for y in range(cols):
                if not grid[x][y]: 
                    if distance[x][y] == float('inf'):
                        neighbors = [
                            (x + dx, y + dy)
                            for dx, dy in [(-1, 0), (1, 0), (0, -1), (0, 1)]
                            if 0 <= x + dx < rows and 0 <= y + dy < cols
                        ]

                        if any(distance[nx][ny] == last_distance for nx, ny in neighbors):
                            distance[x][y] = last_distance + 1
                            new_entry = True

        last_distance += 1

    return distance
```

```python
def load_grid(file_path):
    with open(file_path, 'r') as file:
        data = json.load(file)
    
    grid = data['grid']
    goal_x = data['goal_x'] - 1 
    goal_y = data['goal_y'] - 1
    agent_x = data['agent_x'] - 1
    agent_y = data['agent_y'] - 1

    goal = (goal_x, goal_y)
    start = (agent_x, agent_y) 
    
    return grid, start, goal
```
