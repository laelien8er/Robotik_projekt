```python
import json
from collections import deque
```

```python
def wavefront_algorithm(grid, start, goal):
    # anzahl an zeilen und spalten bestimmen
    rows, cols = len(grid), len(grid[0])
    
    # dist-matrix init mit inf
    distance = [[float('inf')] * cols for _ in range(rows)]

    # parentmatric für pfadrekonstruktion, um vorgängerknoten zu speicher
    parent = [[None] * cols for _ in range(rows)]

    # start- und goalkoordinaten extrahieren
    (agent_x, agent_y) = start
    (goal_x, goal_y) = goal

    # prüfen, ob start und goal blockiert sind
    if grid[agent_x][agent_y] or grid[goal_x][goal_y]:
        raise ValueError("Start oder Ziel ist blockiert.")

    # dist zum startpkt gleich 0 setzen
    distance[agent_x][agent_y] = 0

    # queue für die breitensuch vorbereiten und startknoten einfügen
    queue = deque([start])
    
    # breitensuche ausführen 
    while queue:
        x, y = queue.popleft()

        # wenn goal erreicht, abbrechen
        if (x, y) == (goal_x, goal_y):
            break

        # prüfe alle möglichen nachbarn von (x,y) in die 4 richtungen 
        for dx, dy in [(-1,0), (1,0), (0,-1), (0,1)]:
            nx, ny = x+dx, y+dy

            # prüfen, ob nachbar innerhalb des grids und nicht blockiert
            if 0 <= nx < rows and 0 <= ny < cols:
                # wenn nachbar nicht besucht (inf) und nicht blockiert, dann aktualisieren
                if not grid[nx][ny] and distance[nx][ny] == float('inf'):
                    distance[nx][ny] = distance[x][y] + 1
                    parent[nx][ny] = (x, y)
                    queue.append((nx, ny))

    # wenn distanz am goal immer nich inf >> pfad existiert nicht
    if distance[goal_x][goal_y] == float('inf'):
        raise ValueError("Kein Pfad gefunden.")

    # gibt distanz- und parent-arrays zurück, um pfad zu rekonstruieren
    return distance, parent
```

```python
def find_path(parent, start, goal):
    # liste für pfad erstellen
    path = []

    # starte mit goalknoten
    x, y = goal

    # zurückverfolgen bis man beim startknoten ankommen
    while (x, y) != start:
        # aktuellen knoten zum pfad hinzufügen
        path.append((x, y))
        
        # zum parentknoten des aktuellen knotens gehen
        x, y = parent[x][y]

    # zum schluss den startknoten hinzufügen
    path.append(start)
    
    # pfad umkehren >> da aufgebaut von goal zu start
    path.reverse()

    # ferigen pfad zurückgeben
    return path
```

```python
def load_grid(file_path):
    with open(file_path, 'r') as file:
        data = json.load(file)
    
    grid = data['grid']
    goal_x = data['goal_x'] - 1 
    goal_y = data['goal_y'] - 1
    agent_x = data['agent_x'] - 1
    agent_y = data['agent_y'] - 1

    goal = (goal_x, goal_y)
    start = (agent_x, agent_y) 
    
    return grid, start, goal
```

```python
file_path = 'simple_grid_16.json'

grid, start, goal = load_grid(file_path)
print(start, goal)
```

```python
distance, parent = wavefront_algorithm(grid, start, goal)

# pfad rekonstruieren
path = find_path(parent, start, goal)

print("Gefundener Pfad:", path)
print("Distanz zum Ziel:", distance[goal[0]][goal[1]])
```
