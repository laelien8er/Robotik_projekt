```python
def get_actions(path, start):
    # check if path is start to end
    if path[0] != start:
        path.reverse()
        
    actions = [] 
    p_last = path[0]
    for p in path[1:]:
        if p[1] == p_last[1]:
            if p[0] < p_last[0]: 
                actions.append(2) # step up
            elif p[0] > p_last[0]:
                actions.append(3) # step down
        elif p[0] == p_last[0]:
            if p[1] > p_last[1]: 
                actions.append(0) # step right
            elif p[1] < p_last[1]:
                actions.append(1) # step left
        p_last = p
    return actions
```

```python
actions = get_actions(path, start)
```

```python
env = GridEnv(render_mode="human" , map_file=file_path)
observation, info = env.reset()


episode_over = False
while not episode_over:
    for action in actions:
    # action = env.action_space.sample()  # agent policy that uses the observation and info
        observation, reward, terminated, truncated, info = env.step(action)

        episode_over = terminated or truncated

env.close()
```
